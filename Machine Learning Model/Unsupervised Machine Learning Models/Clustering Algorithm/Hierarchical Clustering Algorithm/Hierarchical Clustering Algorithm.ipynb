# Hierarchical Clustering

- It is used to group the unlabeled datasets into a cluster and also known as hierarchical cluster analysis or HCA.
- In this algorithm, we develop the hierarchy of clusters in the form of a tree, and this tree-shaped structure is known as teh dendrogram.

### What is Hierarchical Clustering?
Hierarchical Clustering is an unsupervised learning technique that builds a hierarchy of clusters. It either:
- Aglomerates data points bottoms-up(most common)
- Divides data top down.

Goal: Create a tree like structure(dendrogram) that shows how data points are grouped.

### Types of Hierarchical Clustering
##### 🔹 1. Agglomerative (Bottom-Up) — Most Used
- Start with each data point as its own cluster.
- Repeatedly merge the closest clusters until only one big cluster remains.

##### 🔹 2. Divisive (Top-Down)
- Start with one big cluster.
- Repeatedly split into smaller clusters.



### Working of Agglomerative Clustering (Step-by-Step)
Let’s say you have 5 data points: A, B, C, D, E.

##### Step 1: Compute Distance Matrix
- Calculate the pairwise distance (Euclidean/Manhattan/Cosine).

A	B	C	D	E
A	-				
B		-			
...	...	...	...	...	...

##### Step 2: Merge Closest Points
- Find the pair with the smallest distance.
- Merge them into a single cluster.

##### Step 3: Update Distance Matrix
- Compute new distances between this new cluster and all others based on linkage criteria (explained below).

##### Step 4: Repeat
- Continue merging the closest clusters until 1 cluster remains or until you decide to stop at a specific number of clusters.

##### Linkage Criteria (How to Measure Distance Between Clusters)

| Linkage Type         | Description                                    | When to Use           |
| -------------------- | ---------------------------------------------- | --------------------- |
| **Single Linkage**   | Distance = minimum between points in clusters  | Elongated clusters    |
| **Complete Linkage** | Distance = maximum between points in clusters  | Compact clusters      |
| **Average Linkage**  | Distance = average of all pairwise distances   | Balanced clusters     |
| **Centroid Linkage** | Distance = between centroids                   | Sensitive to outliers |
| **Ward's Method**    | Minimizes increase in total variance (inertia) | Most commonly used    |

## Agglomerative Hierarichal Clustering

- Start with individual points and merge them into clusters
- Bottom-Up

##### Working 
Working (Bottom-Up)
- Start with each data point as its own cluster (n clusters).
- Compute distances between all clusters.
- Merge the two closest clusters.
- Update the distance matrix.
- Repeat steps 2–4 until all points are merged into a single cluster.

 🧠 Agglomerative = Keep grouping until everything is in one group.

##### Distance Between Clusters (Linkage Criteria)

| Method               | Description                                    |
| -------------------- | ---------------------------------------------- |
| **Single Linkage**   | Closest pair (minimum distance)                |
| **Complete Linkage** | Farthest pair (maximum distance)               |
| **Average Linkage**  | Average of all pairs                           |
| **Ward’s Method**    | Minimizes variance within clusters (most used) |


```python
from sklearn.datasets import load_iris
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

X = load_iris().data

# Model
model = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = model.fit_predict(X)

# Dendrogram
Z = linkage(X, method='ward')
dendrogram(Z)
plt.title("Agglomerative Dendrogram")
plt.show()
```

## Divisive Hierarichal Clustering

- Start with one big cluster and split it into smaller ones
- Top-Down

##### Working (Top-Down)
- Start with all data points in one cluster.
- Split the cluster into two subclusters (based on dissimilarity).
- Repeatedly split the largest/most heterogeneous cluster until each point is in its own cluster or a desired number is reached.

🧠 Divisive = Keep dividing until every point is isolated.
